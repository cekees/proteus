# 2D construction of IFE basis functions

**Author:** Darsh Nathawani

$
\begin{align}
\nabla \cdot \left(f_{\alpha} u - \mu_{\alpha} \nabla u \right) &= s \text{ on } K = K_a \cap K_b \\
\left(f_b u - \mu_b \nabla u \right) \cdot \mathbf{n} &= \left(f_a u - \mu_a \nabla u \right) \cdot \mathbf{n} \text{ on } \Gamma = \partial K_a \cap \partial K_b \\
u_b - u_a &= j \text{ on } \Gamma
\end{align}
$

where $\mathbf{n}$ is the outward unit normal on $K_a$. First we handle the jump in the coefficients by constructing a new basis $\left\{ v_i \right\}$ such that

$
v_i = \left\{ \begin{array}{lr} v_{a,i} & \text{ on } K_a \\
v_{b,i} & \text{ on } K_b \\
\end{array}
\right.
$

subject to the constraints

$
\begin{align}
v_i(\mathbf{x}_j) &= \delta_{ij} \\
v_{b,i} &=  v_{a,i} \text{ on } \Gamma\\
\left(f_b v_{b,i} - \mu_b \nabla v_{b,i} \right) \cdot \mathbf{n} &= \left(f_a v_{a,i} - \mu_a \nabla v_{a,i} \right) \cdot \mathbf{n} 
\end{align}
$

Note, the constraints above are in physical space. We can transform these to the reference cell coordinates with cell map $\mathbf{x}(\hat{\mathbf{x}})$ and its Jacobian $\mathbf{J} = \left\{ \frac{\partial x_i}{\partial \hat{x}_j} \right\}$, for example:

$
\begin{align}
v_i(\mathbf{\hat{x}}_j) &= \delta_{ij} \\
v_{b,i} &=  v_{a,i} \text{ on } \hat{\Gamma}\\
\left(f_b v_{b,i} - \mu_b \mathbf{J}^{-t} \hat{\nabla} v_{b,i} \right) \cdot \mathbf{n} &= \left(f_a v_{a,i} - \mu_a \mathbf{J}^{-t} \hat{\nabla} v_{a,i} \right) \cdot \mathbf{n} \text{ on } \hat{\Gamma}
\end{align}
$

In 2D, the last two constraints need further explanation. Since we are building a linear nodal basis on $K$, we enforce the basis continuity at nodes on the cut edges and the flux continuity at the midpoint of the cut edge. Orienting the reference cell so that the cuts are $(\hat{x}_0, 0), (0,\hat{y}_0)$, the cut midpoint is $(\hat{x}_0/2, \hat{y}_0/2)$ and the constraints are

$
\begin{align}
v_{b,i}(\hat{x}_0,0) &=  v_{a,i} (\hat{x}_0,0) \\
v_{b,i}(0,\hat{y}_0) &=  v_{a,i} (0,\hat{y}_0) \\
\left(f_b v_{b,i} - \mu_b \mathbf{J}^{-t}\nabla v_{b,i} \right) \cdot \mathbf{n} &= \left(f_a v_{a,i} - \mu_a \mathbf{J}^{-t} \nabla v_{a,i} \right) \cdot \mathbf{n} \text{ at } (\hat{x}_0/2, \hat{y}_0/2)
\end{align}
$

These 3 constraints supplement the 3 nodal constraints to yield needed 6 total constraints for the two linear basis functions. Flux continuity follows as before for our FE solution $u_h = \sum_i U_i v_i$ we have at $(\hat{x}_0/2,\hat{y}_0/2)$

$
\left(f_b u_h - \mu_b \nabla u_h \right)\cdot \mathbf{n} =\left(f_b \sum_i U_i v_{b,i} - \mu_b \nabla \sum_i U_i v_{b,i} \right)\cdot \mathbf{n} = \sum_i U_i \left(f_b v_{b,i} - \mu_b \nabla v_{b,i} \right) \cdot \mathbf{n}= \sum_i U_i \left(f_a v_{a,i} - \mu_a \nabla v_{a,i} \right) \cdot \mathbf{n} = \left(f_a u_h - \mu_a \nabla u_h \right)\cdot \mathbf{n}
$

As long as $f_{\alpha}$ and $\mu_{\alpha}$ are constant on the cut, using the linearity of the discrete solution we have 

$
\int_{\Gamma} \left(f_b u_h - \mu_b \nabla u_h \right)\cdot \mathbf{n} dS = \int_{\Gamma} \left(f_a u_h - \mu_a \nabla u_h \right)\cdot \mathbf{n} dS
$

The basis is homogeneous: it carries no flux jump. A flux jump $j_f$ enters the method as a load term, not
through the basis.

## Constraints

Both orders use the reference triangle with the cut from $(\hat{x}_0, 0)$ to $(0, \hat{y}_0)$. Side $a$ is the
part that holds the vertex $(0,0)$, i.e. $\hat{x}/\hat{x}_0 + \hat{y}/\hat{y}_0 < 1$, and $\mathbf{n}$ points from
$a$ to $b$. A node takes its value from the side that holds it.

| | P1 (3 + 3 constraints) | P2 (6 + 6 constraints) |
|---|---|---|
| nodes | $(0,0), (0,1), (1,0)$ | $(0,0), (1,0), (0,1), (\tfrac12,0), (\tfrac12,\tfrac12), (0,\tfrac12)$ |
| continuity $v_b = v_a$ | at $(\hat{x}_0,0)$, $(0,\hat{y}_0)$ | at $(\hat{x}_0,0)$, $(0,\hat{y}_0)$, $(\hat{x}_0/2,\hat{y}_0/2)$ |
| flux continuity | $(\mathbf{f}_b-\mathbf{f}_a)\cdot\mathbf{n}\, v - (\mu_b \partial_n v_b - \mu_a \partial_n v_a) = 0$ at the midpoint | $\mu_b \partial_n v_b = \mu_a \partial_n v_a$ at $(\hat{x}_0,0)$, $(0,\hat{y}_0)$ |
| second derivative | | $\mu_b \partial_{nn} v_b = \mu_a \partial_{nn} v_a$ at the midpoint |

Here $\partial_n v = \mathbf{n}\cdot\mathbf{J}^{-t}\hat{\nabla} v$ and $\partial_{nn} v = \mathbf{t}^T \hat{H} v\, \mathbf{t}$ with
$\mathbf{t} = \mathbf{J}^{-1}\mathbf{n}$ (the code's `Jit.T * n`).

```
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
import matplotlib.tri as mtri

x, y = sp.symbols('x y')
half = sp.Rational(1, 2)
MONOMIALS = {1: [1, x, y], 2: [1, x, y, x*y, x**2, y**2]}
NODES = {1: [(0, 0), (0, 1), (1, 0)],
         2: [(0, 0), (1, 0), (0, 1), (half, 0), (half, half), (0, half)]}


def at(v, P):
    return v.subs({x: P[0], y: P[1]})


def constraints(k, X0, Y0, mua, mub, n, Jit, fa, fb, sides):
    """The linear constraints of the order-k IFE basis.

    n, Jit, fa, fb are sympy matrices (numbers or symbols); sides[i] is 'a' or 'b' for node i.
    Returns (equations, coefficient unknowns, nodal-value symbols).
    """
    m = len(MONOMIALS[k])
    ca, cb = sp.symbols(f'a1:{m + 1}'), sp.symbols(f'b1:{m + 1}')
    va = sum(c*p for c, p in zip(ca, MONOMIALS[k]))
    vb = sum(c*p for c, p in zip(cb, MONOMIALS[k]))
    vals = sp.symbols(f'v1:{len(NODES[k]) + 1}')

    def dn(v):
        return (n.T * Jit * sp.Matrix([v.diff(x), v.diff(y)]))[0]

    eqs = [at(va if s == 'a' else vb, P) - val for P, s, val in zip(NODES[k], sides, vals)]
    cut, mid = [(X0, 0), (0, Y0)], (X0/2, Y0/2)
    if k == 1:
        eqs += [at(vb - va, P) for P in cut]
        eqs += [(fb - fa).dot(n)*at(va, mid) - at(mub*dn(vb) - mua*dn(va), mid)]
    else:
        t = Jit.T * n
        def dnn(v):
            return (t.T * sp.hessian(v, (x, y)) * t)[0]
        eqs += [at(vb - va, P) for P in cut + [mid]]
        eqs += [at(mub*dn(vb) - mua*dn(va), P) for P in cut]
        eqs += [at(mub*dnn(vb) - mua*dnn(va), mid)]
    return eqs, list(ca) + list(cb), vals


def ife_basis(k, X0, Y0, mua, mub, Jit=((1, 0), (0, 1)), fa=(0, 0), fb=(0, 0)):
    """Coefficients of the order-k IFE basis for one cut cell.

    Column i holds basis function i: its side-a coefficients, then its side-b coefficients, in MONOMIALS[k].
    The normal is that of the cut in physical space, pointing from a to b.
    """
    Jit = sp.Matrix(Jit)
    g = Jit * sp.Matrix([1/sp.nsimplify(X0), 1/sp.nsimplify(Y0)])
    sides = ['a' if P[0]/X0 + P[1]/Y0 < 1 else 'b' for P in NODES[k]]
    eqs, unknowns, vals = constraints(k, X0, Y0, mua, mub, g/g.norm(), Jit,
                                      sp.Matrix(fa), sp.Matrix(fb), sides)
    A, rhs = sp.linear_eq_to_matrix(eqs, unknowns)
    return np.linalg.solve(np.array(A, dtype=float), np.array(rhs.jacobian(vals), dtype=float))


def monomials(k, X, Y):
    X, Y = np.asarray(X, float), np.asarray(Y, float)
    one = np.ones_like(X)
    return np.array([one, X, Y] if k == 1 else [one, X, Y, X*Y, X**2, Y**2])


def side_values(k, C, X, Y):
    """Every basis function's side-a and side-b polynomial at the points (X, Y): two (nbasis, npoints) arrays."""
    m = len(MONOMIALS[k])
    P = monomials(k, X, Y)
    return C[:m].T @ P, C[m:].T @ P


def basis_values(k, C, X0, Y0, X, Y):
    """The basis functions themselves: the side-a polynomial on side a, the side-b one on side b."""
    A, B = side_values(k, C, X, Y)
    return np.where(np.asarray(X)/X0 + np.asarray(Y)/Y0 < 1, A, B)
```

## P1: closed form

The P1 system is small enough to solve symbolically, for any cut, viscosities, advection, normal and Jacobian.

```
X0s, Y0s, mua_s, mub_s, nx, ny = sp.symbols('x0 y0 ma mb nx ny')
eqs, unknowns, vals = constraints(1, X0s, Y0s, mua_s, mub_s, sp.Matrix([nx, ny]),
                                  sp.Matrix(2, 2, sp.symbols('Jit00 Jit01 Jit10 Jit11')),
                                  sp.Matrix(sp.symbols('fax fay')), sp.Matrix(sp.symbols('fbx fby')),
                                  ['a', 'b', 'b'])
P1_closed_form = sp.solve(eqs, unknowns)
for c in unknowns:
    print(c, '=', sp.factor(P1_closed_form[c]), ';')
```

## Visualization

Side $a$ in red, side $b$ in green, the cut in black.

```
def plot_basis(k, X0, Y0, mua, mub, **kw):
    C = ife_basis(k, X0, Y0, mua, mub, **kw)
    s = np.linspace(0, 1, 61)
    X, Y = (g.ravel() for g in np.meshgrid(s, s))
    keep = X + Y <= 1 + 1e-12
    tri = mtri.Triangulation(X[keep], Y[keep])
    on_a = (X[keep][tri.triangles].mean(1)/X0 + Y[keep][tri.triangles].mean(1)/Y0) < 1
    A, B = side_values(k, C, X[keep], Y[keep])
    nb = C.shape[1]
    fig, axes = plt.subplots(nb // 3, 3, figsize=(12, 4*(nb // 3)), subplot_kw={'projection': '3d'})
    for i, ax in enumerate(np.ravel(axes)):
        for vals, mask, colour in ((A[i], ~on_a, 'r'), (B[i], on_a, 'g')):
            t = mtri.Triangulation(X[keep], Y[keep], tri.triangles)
            t.set_mask(mask)
            ax.plot_trisurf(t, vals, color=colour, alpha=0.7, linewidth=0)
        ax.plot([X0, 0], [0, Y0], [0, 0], 'k', lw=2)
        ax.set_title(f'P{k} basis function {i + 1}')
        ax.view_init(elev=25, azim=-60)
    fig.suptitle(f'P{k} IFE basis, cut ({X0}, {Y0}), mu = ({mua}, {mub})')
    plt.tight_layout()
    plt.show()


plot_basis(1, 0.75, 0.45, 1.0, 10.0)
plot_basis(2, 0.4, 0.4, 1.0, 10.0)
```

## Tests

Every case below is checked for both orders, over cuts that put the mid-edge nodes on either side, viscosity
pairs from equal to $10^3$ apart (and the negative pair used earlier in this notebook), and an identity and a
sheared Jacobian. Each test asserts; a run that finishes has passed.

1. **Delta property:** $v_i(\mathbf{x}_j) = \delta_{ij}$, each node read from its own side.
2. **Continuity along the whole cut:** $v_{a,i} = v_{b,i}$ at points all along $\Gamma$, not only where it was imposed.
3. **Partition of unity:** $\sum_i v_i \equiv 1$ on both sides.
4. **Equal viscosities:** with $\mu_a = \mu_b$ the IFE basis is the standard Lagrange basis.
5. **Exact solutions are reproduced:** with $\phi = \hat{x}/\hat{x}_0 + \hat{y}/\hat{y}_0 - 1$ (zero on the cut),
   $u = \phi/\mu$ for P1, and $u = (\phi + \phi^2)/\mu$ for P2, are continuous, have continuous flux and (for P2)
   continuous $\mu\,\partial_{nn}u$. They are in the IFE space, so interpolating them at the nodes must give them
   back exactly on both sides. For P1 this also holds with advection, since $u = 0$ on the cut.

```
CUTS = [(0.4, 0.4), (0.75, 0.45), (0.8, 0.4), (0.4, 0.8), (0.8, 0.8), (0.1, 0.9), (0.95, 0.05)]
VISCOSITIES = [(1.0, 1.0), (1.0, 10.0), (10.0, 1.0), (1.25, -3.0), (1.0, 1000.0), (1000.0, 1.0)]
JACOBIANS = {'identity': ((1, 0), (0, 1)), 'sheared': ((2.0, 0.5), (-0.3, 1.5))}
ADVECTION = {'none': ((0, 0), (0, 0)), 'P1 only': ((0.3, -0.2), (1.0, 0.5))}
TOL = 1e-9

rng = np.random.default_rng(0)
s = rng.random((2, 200))
XR, YR = np.minimum(s[0], 1 - s[1]), np.minimum(s[1], 1 - s[0])     # random points in the triangle


def lagrange(k):
    """Standard Lagrange basis coefficients, in the layout of ife_basis with the same polynomial on both sides."""
    V = np.array([[float(at(sp.sympify(p), P)) for p in MONOMIALS[k]] for P in NODES[k]])
    C = np.linalg.inv(V)
    return np.vstack([C, C])


def check(k, X0, Y0, mua, mub, Jit, fa, fb):
    C = ife_basis(k, X0, Y0, mua, mub, Jit=Jit, fa=fa, fb=fb)
    scale = max(1.0, np.abs(C).max())
    err = {}
    # 1. delta property
    NX = np.array([float(P[0]) for P in NODES[k]]); NY = np.array([float(P[1]) for P in NODES[k]])
    err['delta'] = np.abs(basis_values(k, C, X0, Y0, NX, NY) - np.eye(len(NX))).max()
    # 2. continuity along the cut
    s = np.linspace(0, 1, 11)
    A, B = side_values(k, C, X0*(1 - s), Y0*s)
    err['continuity'] = np.abs(A - B).max() / scale
    no_advection = np.allclose(fa, 0) and np.allclose(fb, 0)
    if no_advection:
        # 3. partition of unity, on each side's whole polynomial
        A, B = side_values(k, C, XR, YR)
        err['partition of unity'] = max(np.abs(A.sum(0) - 1).max(), np.abs(B.sum(0) - 1).max()) / scale
        # 4. equal viscosities give the standard basis
        if mua == mub:
            err['standard basis'] = np.abs(C - lagrange(k)).max()
    # 5. exact reproduction
    def exact(X, Y, mu):
        phi = np.asarray(X)/X0 + np.asarray(Y)/Y0 - 1
        return (phi if k == 1 else phi + phi**2) / mu
    on_a = NX/X0 + NY/Y0 < 1
    U = np.where(on_a, exact(NX, NY, mua), exact(NX, NY, mub))
    A, B = side_values(k, C, XR, YR)
    err['reproduction'] = max(np.abs(U @ A - exact(XR, YR, mua)).max(),
                              np.abs(U @ B - exact(XR, YR, mub)).max()) / max(1.0, np.abs(U).max())
    return err


worst = {}
cases = 0
for k in (1, 2):
    for (X0, Y0) in CUTS:
        for (mua, mub) in VISCOSITIES:
            for jname, Jit in JACOBIANS.items():
                for aname, (fa, fb) in ADVECTION.items():
                    if k == 2 and aname != 'none':
                        continue
                    for test, e in check(k, X0, Y0, mua, mub, Jit, fa, fb).items():
                        key = (k, test)
                        if e > worst.get(key, (-1,))[0]:
                            worst[key] = (e, (X0, Y0), (mua, mub), jname, aname)
                    cases += 1

print(f'{cases} cut cells checked')
print(f"{'order':6s} {'test':20s} {'worst error':>12s}   worst case")
for (k, test), (e, cut, mu, jname, aname) in sorted(worst.items()):
    print(f'P{k:<5d} {test:20s} {e:12.1e}   cut {cut}, mu {mu}, {jname} Jacobian, advection {aname}')
failed = {key: v for key, v in worst.items() if v[0] > TOL}
assert not failed, f'tests above {TOL:g}: {failed}'
```
